```python

KB = ["Human(Ram)", "Human(Sita)"]

rules = {
    "Human": "Mortal"
}

changed = True

while changed:
    changed = False

    for fact in KB[:]:
        for condition, conclusion in rules.items():

            if fact.startswith(condition):
                name = fact[len(condition) + 1:-1]
                new_fact = conclusion + "(" + name + ")"

                if new_fact not in KB:
                    KB.append(new_fact)
                    changed = True

print("Knowledge Base:")

for fact in KB:
    print(fact)


```
