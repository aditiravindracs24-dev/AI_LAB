```python
def get_moves(state):
    pos = state.index(0)
    r, c = pos // 3, pos % 3
    for dr, dc in [(-1,0), (1,0), (0,-1), (0,1)]:
        if 0 <= r+dr < 3 and 0 <= c+dc < 3:
            new_pos = (r+dr)*3 + (c+dc)
            board = list(state)
            board[pos], board[new_pos] = board[new_pos], board[pos]
            yield tuple(board)

def dfs(state, goal, path, visited):
    if state == goal: return path
    if len(path) > 10: return None
    visited.add(state)
    for nxt in get_moves(state):
        if nxt not in visited:
            res = dfs(nxt, goal, path + [nxt], visited)
            if res: return res
    visited.remove(state)
    return None

start, goal = (1, 2, 3, 0, 4, 6, 7, 5, 8), (1, 2, 3, 4, 5, 6, 7, 8, 0)
sol = dfs(start, goal, [start], set())

for step in sol or []:
    print(step[0:3], '\n', step[3:6], '\n', step[6:9], '\n---')
```

```python
def dls(state, goal, path, visited, depth_limit):
    if state == goal: return path
    if len(path) - 1 >= depth_limit: return None

    visited.add(state)
    for nxt in get_moves(state):
        if nxt not in visited:
            res = dls(nxt, goal, path + [nxt], visited, depth_limit)
            if res: return res
    visited.remove(state)
    return None

def idfs(start, goal):
    depth = 0
    while True:
        res = dls(start, goal, [start], set(), depth)
        if res: return res, depth
        depth += 1

sol, reached_depth = idfs(start, goal)
print(f"Solved using IDFS at depth {reached_depth}!")
for step in sol:
    print(step[0:3], '\n', step[3:6], '\n', step[6:9], '\n---')
```

```python
import heapq

def manhattan_distance(state, goal):
    distance = 0
    for i in range(9):
        val = state[i]
        if val != 0:
            r, c = i // 3, i % 3
            target_idx = goal.index(val)
            tr, tc = target_idx // 3, target_idx % 3
            distance += abs(r - tr) + abs(c - tc)
    return distance

def a_star(start, goal):
    queue = [(manhattan_distance(start, goal), start, [start])]
    visited = set()

    while queue:
        f_score, current, path = heapq.heappop(queue)

        if current == goal:
            return path

        if current in visited:
            continue
        visited.add(current)

        for nxt in get_moves(current):
            if nxt not in visited:
                g_score = len(path)
                h_score = manhattan_distance(nxt, goal)
                heapq.heappush(queue, (g_score + h_score, nxt, path + [nxt]))
    return None

sol = a_star(start, goal)
print(f"Solved using A* in {len(sol) - 1} steps!")
for step in sol:
    print(step[0:3], '\n', step[3:6], '\n', step[6:9], '\n---')
```
