```python
import random

N = 4


def calculate_cost(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts



def get_neighbors(state):
    neighbors = []

    for col in range(N):
        for row in range(N):
            if row != state[col]:
                new_state = state.copy()
                new_state[col] = row
                neighbors.append(new_state)

    return neighbors


def hill_climbing():

    current = [0, 1, 2, 3]
    random.shuffle(current)

    while True:
        current_cost = calculate_cost(current)

        print("\nCurrent State:", current)
        print("Current Cost:", current_cost)

        if current_cost == 0:
            print("\nGoal State Reached!")
            return current

        neighbors = get_neighbors(current)


        best_neighbor = None
        best_cost = current_cost

        print("\nNeighbor Costs:")

        for neighbor in neighbors:
            cost = calculate_cost(neighbor)
            print(neighbor, "Cost =", cost)

            if cost < best_cost:
                best_cost = cost
                best_neighbor = neighbor


        if best_neighbor is None:
            print("\nLocal Minimum Reached!")
            return None

        current = best_neighbor


solution = hill_climbing()

if solution:
    print("\nFinal Solution:", solution)
    print("Final Cost:", calculate_cost(solution))
```

```python
import random
import math

N = 8


def calculate_cost(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):


            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts



def get_neighbor(state):
    new_state = state.copy()


    col = random.randint(0, N - 1)


    new_row = random.randint(0, N - 1)

    while new_row == new_state[col]:
        new_row = random.randint(0, N - 1)

    new_state[col] = new_row

    return new_state



def simulated_annealing():


    current = [random.randint(0, N - 1)
               for _ in range(N)]

    current_cost = calculate_cost(current)


    temperature = 100.0


    cooling_rate = 0.99

    iteration = 0

    while temperature > 0.001:

        if current_cost == 0:
            print("\nGoal State Reached!")
            break


        neighbor = get_neighbor(current)

        neighbor_cost = calculate_cost(neighbor)


        delta = neighbor_cost - current_cost


        if delta < 0:
            current = neighbor
            current_cost = neighbor_cost


        else:
            probability = math.exp(-delta / temperature)

            if random.random() < probability:
                current = neighbor
                current_cost = neighbor_cost


        temperature *= cooling_rate

        iteration += 1

        if iteration % 100 == 0:
            print("Iteration:", iteration,
                  "Cost:", current_cost,
                  "Temperature:", round(temperature, 4))

    print("\nFinal Board:", current)
    print("Final Cost:", current_cost)

    if current_cost == 0:
        print("8-Queens Solution Found!")
    else:
        print("Solution Not Found. Try Again.")

    return current



simulated_annealing()
```
